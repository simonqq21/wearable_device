# Serial Receiver

```python
import serial
import io
import numpy as np 
import os
from pathlib import Path
import struct
from cobs import cobs
```

```python
ser = serial.Serial('/dev/ttyUSB0')
print(f"ser.name = {ser.name}")
```

# Define orientation data struct

```python
USE_EULER =  True
USE_QUATERNION = False 

if USE_EULER:
    # 1 uint64_t, 3 floats
    STRUCT_FORMAT = "<Q4f" 
elif USE_QUATERNION: 
    # 1 uint64_t, 4 floats
    STRUCT_FORMAT = "<Q4f"

struct_size = struct.calcsize(STRUCT_FORMAT) 
print(f"struct_size: {struct_size} bytes")
```

# Open serial port and 

```python
while True:
    with serial.Serial('/dev/ttyUSB0', 230400) as ser:
        in_data = ser.read_until(b'\0')
        # print(f"in_data = {in_data}") 
        
        # ser.read_until(b'0')
        # ser.read(1)
        
        in_data = b'' 
        while (len(in_data) < struct_size + 2):
            chunk = ser.read(struct_size + 2 - len(in_data))
            # print(f"len(chunk) = {len(chunk)}")
            in_data += chunk;
            
        # ser.read(1)
        # print(f"in_data = {in_data}") 
        # print(f"len(in_data) = {len(in_data)}")
        try:
            decoded = cobs.decode(in_data[:-1])
            # print(f"len(decoded) = {len(decoded)}")
            if (len(decoded) == struct_size):
                # Unpack the raw bytes into Python variables
                try:
                    unpacked_data = struct.unpack(STRUCT_FORMAT, decoded)
                    timestamp = unpacked_data[0] 
                    if USE_EULER:
                        x = unpacked_data[1]
                        y = unpacked_data[2] 
                        z = unpacked_data[3]
                        print(f"timestamp = {timestamp}, x = {x}, y = {y}, z = {z}")
                    elif USE_QUATERNION:
                        w = unpacked_data[1]
                        x = unpacked_data[2] 
                        y = unpacked_data[3]
                        z = unpacked_data[4]
                        print(f"timestamp = {timestamp}, w = {w}, x = {x}, y = {y}, z = {z}")
                except:
                    pass
        except:
            pass
```
